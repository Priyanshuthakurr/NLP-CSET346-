```python
!pip install -q nltk pandas

import nltk
import pandas as pd

from nltk.corpus import wordnet as wn
from nltk.corpus import wordnet_ic

nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('wordnet_ic')

print("Libraries and WordNet loaded successfully!")
```

```python
words = [
    "car",
    "bank",
    "bat",
    "dog",
    "light",
    "plant",
    "book",
    "teacher",
    "student",
    "computer"
]

print("Selected Words:")
print(words)
```

```python

q1_data = []

for word in words:
    synsets = wn.synsets(word)

    print("\n" + "=" * 70)
    print("WORD:", word)
    print("Number of Synsets:", len(synsets))

    for synset in synsets:
        definition = synset.definition()
        examples = synset.examples()

        print("\nSynset:", synset.name())
        print("Definition:", definition)

        if examples:
            print("Example:", examples[0])
        else:
            print("Example: No example available")

        q1_data.append({
            "Word": word,
            "Synset": synset.name(),
            "Definition": definition,
            "Example": examples[0] if examples else "No example available"
        })


q1_df = pd.DataFrame(q1_data)

print("\n\nQ1 DATAFRAME:")
display(q1_df)
```

```python

print("Words with Multiple Meanings:\n")

for word in words:
    synsets = wn.synsets(word)

    if len(synsets) >= 3:
        print("-" * 60)
        print("Word:", word)


        for i, synset in enumerate(synsets[:3], start=1):
            print(f"Meaning {i}:")
            print("  Synset:", synset.name())
            print("  Definition:", synset.definition())
```

```python

nouns = [
    "dog",
    "cat",
    "car",
    "apple",
    "computer",
    "teacher",
    "student",
    "bird",
    "tree",
    "book"
]

print("Selected Nouns:")
print(nouns)
```

```python

q2_data = []

for word in nouns:
    synsets = wn.synsets(word, pos=wn.NOUN)

    if synsets:
        synset = synsets[0]
        hypernyms = synset.hypernyms()

        direct_hypernyms = [h.name() for h in hypernyms]

        q2_data.append({
            "Word": word,
            "Synset": synset.name(),
            "Direct Hypernyms": ", ".join(direct_hypernyms)
        })

q2_df = pd.DataFrame(q2_data)

print("Q2: Direct Hypernyms")
display(q2_df)
```

```python

def get_hypernym_chain(synset, levels=3):
    chain = []
    current = synset

    for level in range(levels):
        if current is None:
            break

        chain.append(current.name())

        hypernyms = current.hypernyms()

        if hypernyms:
            current = hypernyms[0]
        else:
            break

    return chain


hierarchy_words = ["dog", "cat", "car", "bird", "apple"]

q2_hierarchy = []

for word in hierarchy_words:
    synsets = wn.synsets(word, pos=wn.NOUN)

    if synsets:
        synset = synsets[0]
        chain = get_hypernym_chain(synset, levels=4)

        q2_hierarchy.append({
            "Word": word,
            "Level 0": chain[0] if len(chain) > 0 else "",
            "Level 1": chain[1] if len(chain) > 1 else "",
            "Level 2": chain[2] if len(chain) > 2 else "",
            "Level 3": chain[3] if len(chain) > 3 else ""
        })

hierarchy_df = pd.DataFrame(q2_hierarchy)

print("Q2: Hypernym Hierarchy")
display(hierarchy_df)
```

```python

word_pairs = [
    ("car", "automobile"),
    ("dog", "cat"),
    ("dog", "animal"),
    ("dog", "computer"),
    ("car", "vehicle"),
    ("cat", "animal"),
    ("book", "computer"),
    ("teacher", "student")
]

print("Selected Word Pairs:")
for pair in word_pairs:
    print(pair)
```

```python

similarity_data = []

for word1, word2 in word_pairs:

    synsets1 = wn.synsets(word1, pos=wn.NOUN)
    synsets2 = wn.synsets(word2, pos=wn.NOUN)

    if synsets1 and synsets2:


        max_similarity = 0
        best_synset1 = None
        best_synset2 = None

        for s1 in synsets1:
            for s2 in synsets2:
                similarity = s1.path_similarity(s2)

                if similarity is not None and similarity > max_similarity:
                    max_similarity = similarity
                    best_synset1 = s1
                    best_synset2 = s2

        similarity_data.append({
            "Word 1": word1,
            "Word 2": word2,
            "Synset 1": best_synset1.name() if best_synset1 else "N/A",
            "Synset 2": best_synset2.name() if best_synset2 else "N/A",
            "Similarity Score": max_similarity
        })

q3_df = pd.DataFrame(similarity_data)

print("Q3: Semantic Similarity")
display(q3_df)
```

```python

q3_ranked = q3_df.sort_values(
    by="Similarity Score",
    ascending=False
).reset_index(drop=True)

q3_ranked.insert(0, "Rank", range(1, len(q3_ranked) + 1))

print("Ranked Semantic Similarity:")
display(q3_ranked)
```

```python
print("Q3 Observations:\n")

for _, row in q3_ranked.iterrows():
    score = row["Similarity Score"]

    if score >= 0.5:
        observation = "Highly semantically related"
    elif score >= 0.2:
        observation = "Moderately semantically related"
    else:
        observation = "Weakly related / semantically distant"

    print(
        f"{row['Word 1']} - {row['Word 2']}: "
        f"{score:.3f} -> {observation}"
    )
```
