```python
# ============================================
# NLP LAB - 1
# Text Preprocessing and N-gram Analysis
# Priyanshu Thakur (S24CSEU2222)
# ============================================

import nltk
import pandas as pd
import string
from collections import Counter
from difflib import get_close_matches

resources = [
    'brown',
    'stopwords',
    'punkt',
    'punkt_tab',
    'words',
    'wordnet',
    'omw-1.4'
]

for resource in resources:
    nltk.download(resource, quiet=True)

print("All NLTK resources downloaded successfully!")
```

```python
from nltk.corpus import brown, stopwords
from nltk.tokenize import sent_tokenize, word_tokenize


brown_sentences = brown.sents()


selected_sentences = [' '.join(sentence) for sentence in brown_sentences[:20]]

print("Number of selected sentences:", len(selected_sentences))
print("\nFirst 5 Original Sentences:\n")

for i, sentence in enumerate(selected_sentences[:5], 1):
    print(f"{i}. {sentence}")
```

```python

text = " ".join(selected_sentences)

sentence_tokens = sent_tokenize(text)


word_tokens = word_tokenize(text)

print("Total Sentence Tokens:", len(sentence_tokens))
print("Total Word Tokens:", len(word_tokens))

print("\nSentence Tokens:")
print(sentence_tokens[:5])

print("\nWord Tokens:")
print(word_tokens[:30])
```

```python

stop_words = set(stopwords.words('english'))


lower_tokens = [word.lower() for word in word_tokens]


cleaned_tokens = [
    word for word in lower_tokens
    if word.isalpha() and word not in stop_words
]

print("Original number of tokens:", len(word_tokens))
print("Cleaned number of tokens:", len(cleaned_tokens))

print("\nCleaned Tokens:")
print(cleaned_tokens[:50])
```

```python
print("=" * 80)
print("ORIGINAL AND CLEANED SENTENCES")
print("=" * 80)

for i in range(5):
    original = selected_sentences[i]


    tokens = word_tokenize(original.lower())

    cleaned = [
        word for word in tokens
        if word.isalpha() and word not in stop_words
    ]

    print(f"\nSentence {i+1}")
    print("Original :", original)
    print("Tokens   :", tokens)
    print("Cleaned  :", cleaned)
```

```python
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.corpus import brown


all_words = [
    word.lower()
    for word in brown.words()
    if word.isalpha()
]


selected_words = all_words[:30]


required_words = [
    "playing",
    "studies",
    "running",
    "better",
    "cars"
]

selected_words = selected_words + required_words


stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()


pos_map = {
    "playing": "v",
    "studies": "n",
    "running": "v",
    "better": "a",
    "cars": "n"
}


data = []

for word in selected_words:
    stem = stemmer.stem(word)

    if word in pos_map:
        lemma = lemmatizer.lemmatize(word, pos=pos_map[word])
    else:
        lemma = lemmatizer.lemmatize(word)

    data.append([word, stem, lemma])

df = pd.DataFrame(
    data,
    columns=["Original Word", "Stemmed Word", "Lemmatized Word"]
)

print("Stemming vs Lemmatization:")
display(df)
```

```python
print("Required Examples:")
display(
    df[df["Original Word"].isin(
        ["playing", "studies", "running", "better", "cars"]
    )]
)
```

```python
from nltk.corpus import words

word_vocabulary = set(
    word.lower()
    for word in words.words()
    if word.isalpha()
)

print("Total vocabulary words:", len(word_vocabulary))

misspelled_words = [
    "recieve",
    "teh",
    "definately",
    "seperate",
    "enviroment",
    "becuase",
    "langauge",
    "computar",
    "studing",
    "progrmming"
]

print("\nMisspelled Words:")
print(misspelled_words)
```

```python
def correct_spelling(word, vocabulary, n=3):
    """
    Find the closest words from the NLTK vocabulary.
    """
    suggestions = get_close_matches(
        word.lower(),
        vocabulary,
        n=n,
        cutoff=0.60
    )

    return suggestions



spelling_results = []

for word in misspelled_words:
    suggestions = correct_spelling(word, word_vocabulary)

    spelling_results.append([
        word,
        ", ".join(suggestions) if suggestions else "No suggestion"
    ])

spelling_df = pd.DataFrame(
    spelling_results,
    columns=["Misspelled Word", "Suggested Corrections"]
)

display(spelling_df)
```

```python

example_sentences = [
    "I recieve many messages every day.",
    "Teh student is studying NLP.",
    "This is definately a useful subject.",
    "We need to seperate the words.",
    "The enviroment affects our learning.",
    "I missed the class becuase I was sick."
]

correction_map = {
    "recieve": "receive",
    "teh": "the",
    "definately": "definitely",
    "seperate": "separate",
    "enviroment": "environment",
    "becuase": "because"
}

corrected_sentences = []

for sentence in example_sentences:
    corrected = sentence

    for wrong, correct in correction_map.items():
        corrected = corrected.replace(wrong, correct)
        corrected = corrected.replace(
            wrong.capitalize(),
            correct.capitalize()
        )

    corrected_sentences.append(corrected)

print("=" * 80)
print("SPELLING CORRECTION - SENTENCE EXAMPLES")
print("=" * 80)

for i in range(len(example_sentences)):
    print(f"\nExample {i+1}")
    print("Original :", example_sentences[i])
    print("Corrected:", corrected_sentences[i])
```

```python
from nltk.util import ngrams


brown_words = brown.words()

clean_brown_tokens = [
    word.lower()
    for word in brown_words
    if word.isalpha()
]

print("Total cleaned Brown Corpus tokens:", len(clean_brown_tokens))

print("\nFirst 50 tokens:")
print(clean_brown_tokens[:50])
```

```python

unigrams = list(ngrams(clean_brown_tokens, 1))


bigrams = list(ngrams(clean_brown_tokens, 2))


trigrams = list(ngrams(clean_brown_tokens, 3))

print("Total Unigrams :", len(unigrams))
print("Total Bigrams  :", len(bigrams))
print("Total Trigrams :", len(trigrams))

print("\nSample Unigrams:")
print(unigrams[:10])

print("\nSample Bigrams:")
print(bigrams[:10])

print("\nSample Trigrams:")
print(trigrams[:10])
```

```python

unigram_freq = Counter(unigrams)
bigram_freq = Counter(bigrams)
trigram_freq = Counter(trigrams)


top_unigrams = unigram_freq.most_common(10)
top_bigrams = bigram_freq.most_common(10)
top_trigrams = trigram_freq.most_common(10)

print("TOP 10 UNIGRAMS")
print("-" * 40)

for item, frequency in top_unigrams:
    print(item[0], ":", frequency)

print("\nTOP 10 BIGRAMS")
print("-" * 40)

for item, frequency in top_bigrams:
    print(" ".join(item), ":", frequency)

print("\nTOP 10 TRIGRAMS")
print("-" * 40)

for item, frequency in top_trigrams:
    print(" ".join(item), ":", frequency)
```

```python


unigram_df = pd.DataFrame(
    [(item[0], freq) for item, freq in top_unigrams],
    columns=["Unigram", "Frequency"]
)

bigram_df = pd.DataFrame(
    [(" ".join(item), freq) for item, freq in top_bigrams],
    columns=["Bigram", "Frequency"]
)

trigram_df = pd.DataFrame(
    [(" ".join(item), freq) for item, freq in top_trigrams],
    columns=["Trigram", "Frequency"]
)

print("Top 10 Unigrams")
display(unigram_df)

print("Top 10 Bigrams")
display(bigram_df)

print("Top 10 Trigrams")
display(trigram_df)
```
